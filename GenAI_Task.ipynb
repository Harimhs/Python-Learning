```python
from diffusers import StableDiffusionPipeline
import torch
from google.colab import files
import shutil
import os

uploaded = files.upload()

filename = list(uploaded.keys())[0]
destination_dir = "/content/drive/MyDrive/genai_challenge/data/"
destination_path = os.path.join(destination_dir, "test.jpg")
shutil.copy(filename, destination_path)

print(f"Image saved to: {destination_path}")

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16,
    use_safetensors=True,
).to("cuda")

prompt = "Superman flying in the sky"
image = pipe(prompt).images[0]
image.save("/content/drive/MyDrive/genai_challenge/outputs/images/hero_test.png")
```
